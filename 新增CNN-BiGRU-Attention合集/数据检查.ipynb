```python
import torch
import numpy as np
import pickle

def load_pickle_files():
    """第一步：仅加载数据文件"""

    print("="*60)
    print("模块1：加载原始数据文件")
    print("="*60)

    try:
        # 加载四个文件
        with open('./dataresult/train_set', 'rb') as f:
            train_set = pickle.load(f)

        with open('./dataresult/train_label', 'rb') as f:
            train_label = pickle.load(f)

        with open('./dataresult/test_set', 'rb') as f:
            test_set = pickle.load(f)

        with open('./dataresult/test_label', 'rb') as f:
            test_label = pickle.load(f)

        print("✅ 文件加载成功！")

    except FileNotFoundError as e:
        print(f"❌ 文件未找到: {e}")
        print("请检查文件路径是否正确：")
        print("1. ../dataresult/train_set.pkl")
        print("2. ../dataresult/train_label.pkl")
        print("3. ../dataresult/test_set.pkl")
        print("4. ../dataresult/test_label.pkl")
        return None, None, None, None

    return train_set, train_label, test_set, test_label

# 运行模块1
print("开始模块1：加载数据文件...")
train_set, train_label, test_set, test_label = load_pickle_files()

if train_set is not None:
    print("\n✅ 模块1完成！")
    print("现在进入模块2...")
```

```python
import torch
import numpy as np

def simple_data_check(train_set, train_label, test_set, test_label):
    """
    简单数据检查 - 只输出关键信息
    """
    print("=" * 50)
    print("简单数据检查结果")
    print("=" * 50)

    # 1. 基本信息
    print("\n📊 1. 数据形状:")
    print(f"train_set: {train_set.shape}")
    print(f"train_label: {train_label.shape}")
    print(f"test_set: {test_set.shape}")
    print(f"test_label: {test_label.shape}")

    # 转换为numpy
    if isinstance(train_set, torch.Tensor):
        X_train = train_set.numpy()
        y_train = train_label.numpy()
        X_test = test_set.numpy()
        y_test = test_label.numpy()
    else:
        X_train = np.array(train_set)
        y_train = np.array(train_label)
        X_test = np.array(test_set)
        y_test = np.array(test_label)

    # 2. 特征统计
    print("\n📈 2. 特征统计:")
    n_features = X_train.shape[2]

    for i in range(n_features):
        feature_data = X_train[:, :, i].flatten()
        mean_val = np.mean(feature_data)
        std_val = np.std(feature_data)
        min_val = np.min(feature_data)
        max_val = np.max(feature_data)

        print(f"特征{i:2d}: 均值={mean_val:8.4f}, 标准差={std_val:8.4f}, 范围=[{min_val:8.4f}, {max_val:8.4f}]")

    # 3. 标签统计
    print("\n🎯 3. 标签统计:")
    print(f"训练集标签: 均值={np.mean(y_train):.6f}, 标准差={np.std(y_train):.6f}, 范围=[{np.min(y_train):.6f}, {np.max(y_train):.6f}]")
    print(f"测试集标签: 均值={np.mean(y_test):.6f}, 标准差={np.std(y_test):.6f}, 范围=[{np.min(y_test):.6f}, {np.max(y_test):.6f}]")

    # 4. 数据质量
    print("\n🔍 4. 数据质量检查:")
    print(f"训练集NaN值: {np.isnan(X_train).sum()}/{X_train.size}")
    print(f"训练集Inf值: {np.isinf(X_train).sum()}")
    print(f"测试集NaN值: {np.isnan(X_test).sum()}/{X_test.size}")

    print("\n✅ 检查完成！请将以上结果复制给我分析。")

# 运行检查
if train_set is not None:
    simple_data_check(train_set, train_label, test_set, test_label)
else:
    print("❌ 数据加载失败")
```

```python
import torch
import numpy as np

def export_data_for_analysis(train_set, train_label, test_set, test_label):
    """导出数据用于分析标准化状态"""

    print("="*60)
    print("数据导出分析")
    print("="*60)

    # 确保是Tensor
    if not torch.is_tensor(train_set):
        train_set = torch.tensor(train_set, dtype=torch.float32)
    if not torch.is_tensor(train_label):
        train_label = torch.tensor(train_label, dtype=torch.float32)
    if not torch.is_tensor(test_set):
        test_set = torch.tensor(test_set, dtype=torch.float32)
    if not torch.is_tensor(test_label):
        test_label = torch.tensor(test_label, dtype=torch.float32)

    # 1. 标签统计
    train_label_flat = train_label.flatten()
    test_label_flat = test_label.flatten()

    print("\n1. 标签统计:")
    print(f"训练标签 - 最小值: {train_label_flat.min():.6f}")
    print(f"          最大值: {train_label_flat.max():.6f}")
    print(f"          均值:   {train_label_flat.mean():.6f}")
    print(f"          标准差: {train_label_flat.std():.6f}")

    print(f"\n测试标签 - 最小值: {test_label_flat.min():.6f}")
    print(f"          最大值: {test_label_flat.max():.6f}")
    print(f"          均值:   {test_label_flat.mean():.6f}")
    print(f"          标准差: {test_label_flat.std():.6f}")

    # 2. 特征统计（取前3个特征的最后时间步）
    print("\n2. 特征统计（最后时间步，前3个特征）:")
    feature_names = ['mean', 'std', 'peak', 'rms', 'kurtosis', 'crest', 'shape', 'peak_freq', 'peak_power']

    stats_output = []
    for i in range(min(3, train_set.shape[2])):
        feat_name = feature_names[i] if i < len(feature_names) else f'feature_{i}'

        train_feat = train_set[:, -1, i].flatten()
        test_feat = test_set[:, -1, i].flatten()

        stats = {
            'name': feat_name,
            'train_min': train_feat.min().item(),
            'train_max': train_feat.max().item(),
            'train_mean': train_feat.mean().item(),
            'train_std': train_feat.std().item(),
            'test_min': test_feat.min().item(),
            'test_max': test_feat.max().item(),
            'test_mean': test_feat.mean().item(),
            'test_std': test_feat.std().item()
        }

        stats_output.append(stats)

        print(f"\n{feat_name}:")
        print(f"  训练集 - 范围: [{stats['train_min']:.6f}, {stats['train_max']:.6f}]")
        print(f"           均值: {stats['train_mean']:.6f}, 标准差: {stats['train_std']:.6f}")
        print(f"  测试集 - 范围: [{stats['test_min']:.6f}, {stats['test_max']:.6f}]")
        print(f"           均值: {stats['test_mean']:.6f}, 标准差: {stats['test_std']:.6f}")

    # 3. 导出为可复制格式
    print("\n" + "="*60)
    print("3. 数据摘要（复制给我分析）:")
    print("-" * 60)

    print("标签摘要:")
    print(f"训练标签范围: [{train_label_flat.min():.6f}, {train_label_flat.max():.6f}]")
    print(f"训练标签均值±标准差: {train_label_flat.mean():.6f} ± {train_label_flat.std():.6f}")
    print(f"测试标签范围: [{test_label_flat.min():.6f}, {test_label_flat.max():.6f}]")
    print(f"测试标签均值±标准差: {test_label_flat.mean():.6f} ± {test_label_flat.std():.6f}")

    print("\n特征摘要（前3个）:")
    for stats in stats_output:
        print(f"\n{stats['name']}:")
        print(f"  训练: [{stats['train_min']:.6f}, {stats['train_max']:.6f}] | {stats['train_mean']:.6f} ± {stats['train_std']:.6f}")
        print(f"  测试: [{stats['test_min']:.6f}, {stats['test_max']:.6f}] | {stats['test_mean']:.6f} ± {stats['test_std']:.6f}")

    return stats_output

# 使用示例（在你的数据加载后调用）
print("开始数据导出分析...")
stats = export_data_for_analysis(train_set, train_label, test_set, test_label)
```

```python
# 模块2：检查数据形状和类型
print("\n" + "="*60)
print("模块2：数据形状和类型检查")
print("="*60)

# 转换为Tensor（如果需要）
if not torch.is_tensor(train_set):
    train_set = torch.tensor(train_set, dtype=torch.float32)
if not torch.is_tensor(train_label):
    train_label = torch.tensor(train_label, dtype=torch.float32)
if not torch.is_tensor(test_set):
    test_set = torch.tensor(test_set, dtype=torch.float32)
if not torch.is_tensor(test_label):
    test_label = torch.tensor(test_label, dtype=torch.float32)

print("数据形状:")
print(f"训练集特征: {train_set.shape}")
print(f"训练集标签: {train_label.shape}")
print(f"测试集特征: {test_set.shape}")
print(f"测试集标签: {test_label.shape}")

print("\n数据类型:")
print(f"训练特征: {train_set.dtype}")
print(f"训练标签: {train_label.dtype}")

print("\n✅ 模块2完成！")
print("现在进入模块3...")
```

```python
# 模块3：检查标签（剩余寿命）分布
print("\n" + "="*60)
print("模块3：标签分布检查")
print("="*60)

# 标签范围
train_label_flat = train_label.flatten()
test_label_flat = test_label.flatten()

print("1. 标签范围:")
print(f"   训练集: [{train_label_flat.min():.4f}, {train_label_flat.max():.4f}]")
print(f"   测试集: [{test_label_flat.min():.4f}, {test_label_flat.max():.4f}]")

print("\n2. 标签统计:")
print(f"   训练集 - 均值: {train_label_flat.mean():.4f}, 标准差: {train_label_flat.std():.4f}")
print(f"   测试集 - 均值: {test_label_flat.mean():.4f}, 标准差: {test_label_flat.std():.4f}")

print("\n3. 检查是否线性递减（取前5个和后5个）:")
print(f"   训练集前5个: {train_label_flat[:5].numpy().round(4)}")
print(f"   训练集后5个: {train_label_flat[-5:].numpy().round(4)}")
print(f"   测试集前5个: {test_label_flat[:5].numpy().round(4)}")
print(f"   测试集后5个: {test_label_flat[-5:].numpy().round(4)}")

print("\n✅ 模块3完成！")
print("现在进入模块4...")
```

```python
# 模块4：特征-标签相关性分析
print("\n" + "="*60)
print("模块4：特征-标签相关性分析（带注释说明）")
print("="*60)

# 特征名称（根据你的数据）
feature_names = ['mean', 'std', 'peak', 'rms', 'kurtosis', 'crest', 'shape', 'peak_freq', 'peak_power']

# 注意：这里有时间偏移！
# 对于每个窗口：
# 特征 = 时间点[t, t+1, ..., t+9] 的10个数据点
# 标签 = 时间点[t+10] 的剩余寿命
# 所以特征中的任何时间点都不直接对应标签的时间点

print("【重要说明】")
print("特征窗口: 时间点[t]到[t+9]的数据")
print("预测标签: 时间点[t+10]的剩余寿命")
print("特征和标签有时间偏移（差1个时间点）")
print("因此相关性可能不是最高值")
print("-" * 40)

# 方法1：取整个窗口的时间平均（考虑所有时间点信息）
print("\n方法1：整个窗口时间平均特征")
print("（10个时间点的平均值 vs 第11个时间点的剩余寿命）")
print("特征       训练集r    测试集r    一致性    说明")

train_features_avg = train_set.mean(dim=1)  # (796, 9) - 时间维度上平均
test_features_avg = test_set.mean(dim=1)    # (311, 9)

for i in range(min(9, train_features_avg.shape[1])):
    feat_name = feature_names[i] if i < len(feature_names) else f'特征{i}'

    # 计算训练集相关性
    train_corr = torch.corrcoef(torch.stack([
        train_features_avg[:, i], train_label_flat
    ]))[0, 1].item()

    # 计算测试集相关性
    test_corr = torch.corrcoef(torch.stack([
        test_features_avg[:, i], test_label_flat
    ]))[0, 1].item()

    # 判断一致性
    if train_corr * test_corr < 0:
        consistency = "⚠️相反"
        note = "训练和测试趋势相反，严重问题！"
    elif abs(train_corr - test_corr) > 0.3:
        consistency = "⚠️差异大"
        note = "相关性强度差异大"
    elif abs(train_corr) > 0.5 and abs(test_corr) > 0.5:
        consistency = "✓强相关"
        note = "特征与未来寿命强相关"
    elif abs(train_corr) > 0.3 or abs(test_corr) > 0.3:
        consistency = "~中等"
        note = "有一定相关性"
    else:
        consistency = "弱"
        note = "相关性弱"

    print(f"{feat_name:10s} {train_corr:7.3f}    {test_corr:7.3f}    {consistency:6s}    {note}")

print("\n" + "-" * 40)
print("方法2：最后时间步特征（索引9 = 时间点t+9）")
print("（第10个时间点的数据 vs 第11个时间点的剩余寿命）")
print("特征       训练集r    测试集r")

train_features_last = train_set[:, -1, :]  # (796, 9) - 最后时间步
test_features_last = test_set[:, -1, :]    # (311, 9)

for i in range(min(3, train_features_last.shape[1])):  # 只看前3个
    feat_name = feature_names[i] if i < len(feature_names) else f'特征{i}'

    train_corr_last = torch.corrcoef(torch.stack([
        train_features_last[:, i], train_label_flat
    ]))[0, 1].item()

    test_corr_last = torch.corrcoef(torch.stack([
        test_features_last[:, i], test_label_flat
    ]))[0, 1].item()

    print(f"{feat_name:10s} {train_corr_last:7.3f}    {test_corr_last:7.3f}")

print("\n【相关性解读指南】")
print("r > 0.7: 强相关      0.3-0.7: 中等相关      <0.3: 弱相关")
print("正值: 特征越大，剩余寿命越大（正相关）")
print("负值: 特征越大，剩余寿命越小（负相关，对于振动通常如此）")
print("训练集和测试集应该符号相同，数值相近")

print("\n✅ 模块4完成！")
print("现在进入模块5：数据质量问题检查...")
```
